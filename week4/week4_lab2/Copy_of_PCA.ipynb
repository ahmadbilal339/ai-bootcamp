#Principal Component Analysis (Dimentionality Reduction)
**Principal Component Analysis** is a dimensionality reduction technique used in data science and machine learning. It transforms a high-dimensional dataset into a lower-dimensional one by:

1. Identifying the principal components (directions of maximum variance)

2. Projecting the data onto these components

3. Preserving as much information (variance) as possible

```python
import pandas as pd
from sklearn.model_selection import train_test_split
from sklearn.preprocessing import StandardScaler
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LogisticRegression
from sklearn.svm import SVC
from sklearn.ensemble import RandomForestClassifier
from sklearn.decomposition import PCA
from sklearn.model_selection import cross_val_score
```

##Load dataset

```python
df = pd.read_csv("heart.csv")
df
```

# 1st Way

```python
s_dum = pd.get_dummies(df.Sex)
s_dum
```

```python
dum_cp = pd.get_dummies(df.ChestPainType)
dum_cp
```

```python
rest_dum = pd.get_dummies(df.RestingECG)
rest_dum
```

```python
ex_dum = pd.get_dummies(df.ExerciseAngina)
ex_dum
```

```python
st_dum = pd.get_dummies(df.ST_Slope)
st_dum
```

```python
X = pd.concat([df['Age'],s_dum, dum_cp, df.loc[:,['RestingBP','Cholesterol','FastingBS']], rest_dum, df['MaxHR'], ex_dum, df['Oldpeak'], st_dum],axis=1)
X
```

```python
y = df.HeartDisease
y
```

# Alternative Way

```python
df.describe()
```

```python
mean_bp = df.RestingBP.mean()
df['RestingBP'] = df['RestingBP'].replace(0, mean_bp)
df['Cholesterol'] = df['Cholesterol'].replace(0, df.Cholesterol.mean())
df['FastingBS'] = df['FastingBS'].replace(0, df.FastingBS.mean())
df
```

```python
df[df.RestingBP==0]
```

```python
df = pd.get_dummies(df)
df
```

```python
X = df.drop('HeartDisease',axis='columns')
y = df.HeartDisease
print(X)
print(y)
```

# Train Test Split

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.2)
```

# Before PCA

```python
rf = cross_val_score(RandomForestClassifier(n_estimators=35),X_train, y_train)

print(rf)
```

```python
lr = cross_val_score(LogisticRegression(max_iter=5000),X_train, y_train)

print(lr)
```

```python
model = KNeighborsClassifier(n_neighbors=9)
model.fit(X_train, y_train)
model.score(X_test, y_test)
```

# After PCA

## Normalize the Feature Matrix (X) with Standard Scaling


```python
scalar = StandardScaler()
scaled_x = scalar.fit_transform(X)
scaled_x.shape
```

## Perform Dimensionality Reduction Using PCA (95% Variance Retained)


```python
pca = PCA(0.95)
x_pca = pca.fit_transform(scaled_x)
x_pca.shape
```

#Train test split

```python
X_train, X_test, y_train, y_test = train_test_split(x_pca, y, test_size=0.2)
```

##Train on KNN

```python
model = KNeighborsClassifier(n_neighbors=9)
model.fit(X_train, y_train)
model.score(X_test, y_test)
```

##Random Forest

```python
rf = cross_val_score(RandomForestClassifier(n_estimators=35),X_train, y_train)

print(rf)
```

##Logistic Regression

```python
lr = cross_val_score(LogisticRegression(max_iter=150),X_train, y_train)

print(lr)
```

##Try retaining less variance using PCA and determine how it makes a difference.

```python
# PCA retaining 90% variance
pca_90 = PCA(n_components=0.90)
X_pca_90 = pca_90.fit_transform(scaled_x)

print("Original Features :", X.shape[1])
print("Reduced Features  :", X_pca_90.shape[1])

# Train-test split
X_train, X_test, y_train, y_test = train_test_split(
    X_pca_90,
    y,
    test_size=0.2,
    random_state=42
)

# -----------------------------
# KNN
# -----------------------------
knn = KNeighborsClassifier(n_neighbors=9)
knn.fit(X_train, y_train)

knn_score = knn.score(X_test, y_test)

print("\nKNN Accuracy:", knn_score)

# -----------------------------
# Random Forest
# -----------------------------
rf_scores = cross_val_score(
    RandomForestClassifier(n_estimators=35, random_state=42),
    X_train,
    y_train,
    cv=5
)

print("Random Forest CV Accuracy:", rf_scores.mean())

# -----------------------------
# Logistic Regression
# -----------------------------
lr_scores = cross_val_score(
    LogisticRegression(max_iter=5000),
    X_train,
    y_train,
    cv=5
)

print("Logistic Regression CV Accuracy:", lr_scores.mean())
```

## Observation

After reducing the retained variance from 95% to 90%:

- The number of principal components decreased.
- The dimensionality of the dataset was reduced further.
- Model training became faster due to fewer features.
- The model accuracy changed slightly (mention whether it increased, decreased, or remained almost the same based on your output).
- This demonstrates the trade-off between dimensionality reduction and information retention. Retaining less variance removes more information, which may slightly reduce predictive performance while improving computational efficiency.
